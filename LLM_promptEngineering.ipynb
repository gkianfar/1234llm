```python
def correct_col_type(df,col):
    raw_type = str(type(df[col].dtype)).split('.')[-1].split('\'')[0].lower()
    #print(col,raw_type)
    if 'object' in raw_type:
        if 'date' in col or 'timestamp' in col or 'datetime' in col:
            return pd.to_datetime(df[col])
        else:
            return df[col].astype('category')
    else:
        return df[col]


def gen_date_col(df, tcol):
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    df['date'] = df[tcol].dt.date
    return df
```

```python
# Get the unique patient IDs and device types
patients = physiology_df['patient_id'].unique()
device_types = physiology_df['device_type'].unique()

# Create a DataFrame that retains original study days for each patient
study_days = physiology_df[['patient_id', 'date']].drop_duplicates()

# Prepare a list to hold the filled results
filled_results_list = []

# Iterate over each patient
for patient in patients:
    # Get the patient's data
    patient_data = physiology_df[physiology_df['patient_id'] == patient]

    # For each study day, check measurements
    for date in study_days[study_days['patient_id'] == patient]['date']:
        for device in device_types:
            # Get all values for this patient, date, and device type
            values = patient_data.loc[(patient_data['date'] == date) & (patient_data['device_type'] == device), 'value']
            if not values.empty:
                # Compute the average value (considering repetitions)
                avg_value = values.mean()
                unit = patient_data.loc[(patient_data['date'] == date) & (patient_data['device_type'] == device), 'unit'].values[0]
                filled_results_list.append({
                    'patient_id': patient,
                    'date': date,
                    'device_type': device,
                    'value': avg_value,
                    'unit': unit
                })
            else:
                # If there are no measurements, append a None
                filled_results_list.append({
                    'patient_id': patient,
                    'date': date,
                    'device_type': device,
                    'value': None,  # To be filled later
                    'unit': None
                })

# Convert the list of dictionaries to a DataFrame
filled_results = pd.DataFrame(filled_results_list)

# Now, we can fill in the missing values
def fill_missing_values(group):
    for device in device_types:
        # Patient's specific values for this device
        patient_values = group[group['device_type'] == device]['value']

        # If any values exist, calculate the average to fill
        if patient_values.notnull().any():
            avg_patient_value = patient_values.mean()
            group.loc[(group['device_type'] == device) & (group['value'].isnull()), 'value'] = avg_patient_value
        else:
            # If no values exist at all, use the overall average for this device if applicable
            if physiology_df[physiology_df['device_type'] == device]['value'].notna().any():
                overall_avg = physiology_df[physiology_df['device_type'] == device]['value'].mean()
                group.loc[(group['device_type'] == device) & (group['value'].isnull()), 'value'] = overall_avg

    return group

# Apply filling logic to the filled_results DataFrame
filled_physiology_df = filled_results.groupby('patient_id', group_keys=False).apply(fill_missing_values)

# Sort by patient_id and date
filled_physiology_df.sort_values(by=['patient_id', 'date'], inplace=True)
```

```python
def label_outside_all_patients(activity_df, threshold_minutes=2):
    """
    Labels 'Outside' events between successive 'Front Door' and 'Back Door' events
    for all patients and all days in the dataset.
    - Marks both the **start** and **end** of the outside event.
    - Inserts "Outside" events at regular intervals.
    """
    activity_df['timestamp'] = pd.to_datetime(activity_df['timestamp'])  # Ensure datetime format
    activity_df = activity_df.sort_values(by=['patient_id', 'date', 'timestamp'])  # Sort by patient, day, time

    # Container for processed data
    new_data = []

    # Process each patient separately
    for patient_id, patient_group in activity_df.groupby('patient_id'):
        for date, day_group in patient_group.groupby('date'):
            day_group = day_group.sort_values(by='timestamp').reset_index(drop=True)  # Ensure sorted order

            new_rows = []
            for i in range(len(day_group) - 1):
                current_row = day_group.iloc[i]
                next_row = day_group.iloc[i + 1]
                current_time = current_row['timestamp']
                next_time = next_row['timestamp']

                # Check if the time difference exceeds the threshold
                if (next_time - current_time).total_seconds() / 60.0 > threshold_minutes:
                    if (current_row['location_name'] in ['Front Door', 'Back Door']) and \
                       (next_row['location_name'] in ['Front Door', 'Back Door']):

                        # Mark the start of the "Outside" event
                        start_outside = current_row.copy()
                        start_outside['location_name'] = 'Outside'
                        new_rows.append(start_outside)

                        # Insert 'Outside' events at regular intervals
                        temp_time = current_time
                        while (next_time - temp_time).total_seconds() / 60.0 > threshold_minutes:
                            temp_time += pd.Timedelta(minutes=threshold_minutes)
                            if temp_time >= next_time:
                                break
                            new_row = current_row.copy()
                            new_row['timestamp'] = temp_time
                            new_row['location_name'] = 'Outside'
                            new_rows.append(new_row)

                        # Mark the end of the "Outside" event
                        end_outside = next_row.copy()
                        end_outside['location_name'] = 'Outside'
                        new_rows.append(end_outside)

            # Append the modified day_group with new "Outside" events
            new_data.append(pd.concat([day_group, pd.DataFrame(new_rows)], ignore_index=True)
                            .sort_values(by='timestamp').reset_index(drop=True))

    # Combine all processed patient data
    updated_activity_df = pd.concat(new_data, ignore_index=True)

    return updated_activity_df

# Example usage
updated_activity_df = label_outside_all_patients(activity_df)
```

```python

def calculate_mobility_outside_and_location_counts(activity_df):
    """
    Computes:
    - 'Mobility': The number of location transitions per day.
    - 'Outside Time': The total time (in minutes) a patient spent outside per day.
    - Frequency count for each location type (e.g., Bedroom, Kitchen, Lounge).
    """
    # Ensure timestamps are in datetime format
    activity_df['timestamp'] = pd.to_datetime(activity_df['timestamp'])

    # Sort data by patient, date, and time
    activity_df = activity_df.sort_values(by=['patient_id', 'date', 'timestamp'])

    # Remove NaN values from the 'location_name' column
    activity_df = activity_df.dropna(subset=['location_name'])

    # Identify if the location is outside (Front Door or Back Door)
    activity_df['outside'] = activity_df['location_name'].isin(['Front Door', 'Back Door'])

    # Get all unique indoor locations (excluding Front Door & Back Door)
    indoor_locations = sorted(set(activity_df['location_name']) - {'Front Door', 'Back Door'})

    # Create an empty list to store results
    feature_results = []

    # Process each patient separately
    for patient_id, patient_group in activity_df.groupby('patient_id'):
        # Get unique study days for this patient (only days with activity)
        study_days = patient_group['date'].unique()

        # Process each study day
        for study_day in study_days:
            # Get activity data for this specific day
            day_data = patient_group[patient_group['date'] == study_day].copy()

            # Extract location sequence
            location_sequence = day_data['location_name'].values

            # Compute mobility: Count location transitions
            mobility_count = np.sum(np.array(location_sequence[1:]) != np.array(location_sequence[:-1])) if len(location_sequence) > 1 else 0

            # Compute outside time
            total_time_outside = 0
            for i in range(1, len(day_data)):
                if day_data['outside'].iloc[i] and day_data['outside'].iloc[i - 1]:
                    time_diff = (day_data['timestamp'].iloc[i] - day_data['timestamp'].iloc[i - 1]).total_seconds() / 60
                    total_time_outside += time_diff

            # Count occurrences of each indoor location
            location_counts = {f"{loc}_count": (day_data['location_name'] == loc).sum() for loc in indoor_locations}

            # Store result
            result_entry = {
                'patient_id': patient_id,
                'date': study_day,
                'location_change_count': mobility_count,
                'outside_time': round(total_time_outside, 2)  # Round to 2 decimal places
            }
            result_entry.update(location_counts)  # Add location counts dynamically

            feature_results.append(result_entry)

    # Convert results into a DataFrame
    activity_features_df = pd.DataFrame(feature_results)

    return activity_features_df

# Example usage
activity_features_df = calculate_mobility_outside_and_location_counts(activity_df)
print(activity_features_df)
```

```python
def generate_balanced_text_prompts(demographics_df, agit_df, physiology_df, activity_features_df, stats=None):
    """
    Generates structured text prompts for each patient-day using available data.
    Ensures:
    - Uses all available agitated samples
    - Dynamically adjusts the number of normal samples
    - Splits into exactly 250 training samples and 20 test samples (10 Agitated, 10 Normal)
    - Stores test labels for evaluation
    """

    # Convert agitation events into a dictionary for fast lookup
    agit_dict = {(row['patient_id'], row['date']): "Agitated" for _, row in agit_df.iterrows()}

    # Convert demographics into a dictionary for fast lookup
    demographics_dict = demographics_df.set_index('patient_id').to_dict(orient='index')

    # Convert physiology into a dictionary with patient_id & date as keys
    physiology_dict = physiology_df.pivot(index=['patient_id', 'date'], columns='device_type', values='value').to_dict(orient='index')

    # Identify all agitated samples
    agitated_samples = activity_features_df[activity_features_df.apply(lambda row: (row['patient_id'], row['date']) in agit_dict, axis=1)]
    num_agitated = len(agitated_samples)

    # Identify all normal samples (excluding agitated ones)
    normal_samples = activity_features_df[~activity_features_df.apply(lambda row: (row['patient_id'], row['date']) in agit_dict, axis=1)]
    num_normal = len(normal_samples)

    # Adjust selection dynamically
    normal_train_count = min(125, num_normal - 10)  # Ensure at least 10 remain for testing
    normal_test_count = min(10, num_normal - normal_train_count)  # Ensure test is max 10

    agitated_train_count = min(125, num_agitated - 10)  # Ensure at least 10 remain for testing
    agitated_test_count = min(10, num_agitated - agitated_train_count)

    # Select samples dynamically
    train_agitated = agitated_samples.sample(n=agitated_train_count, random_state=42)
    test_agitated = agitated_samples.drop(train_agitated.index).sample(n=agitated_test_count, random_state=42)

    train_normal = normal_samples.sample(n=normal_train_count, random_state=42)
    test_normal = normal_samples.drop(train_normal.index).sample(n=normal_test_count, random_state=42)

    # **Final Dataset Splits**
    train_data = pd.concat([train_agitated, train_normal]).sample(frac=1, random_state=42)  # Shuffle
    test_data = pd.concat([test_agitated, test_normal]).sample(frac=1, random_state=42)  # Shuffle

    # Prepare lists for train and test samples
    train_samples = []
    test_samples = []
    test_labels = []  # Store true labels of test samples

    # Process each patient-day in train and test sets
    for split, dataset, sample_list, mask_label, label_list in zip(
        ["Train", "Test"],
        [train_data, test_data],
        [train_samples, test_samples],
        ["{LABEL}", "[MASK]"],
        [None, test_labels]  # Only store labels for test samples
    ):
        for _, row in dataset.iterrows():
            patient_id = row['patient_id']
            date = row['date']

            # Retrieve demographic information
            demographic_info = demographics_dict.get(patient_id, {})
            age_group = demographic_info.get('age', "Unknown Age Group")
            gender = demographic_info.get('sex', "Unknown Gender")

            # Determine agitation status
            agitation_label = agit_dict.get((patient_id, date), "Normal")

            # Retrieve physiological data for the day
            physiology_data = physiology_dict.get((patient_id, date), {})
            physiology_text = "\n".join([f"- {k}: {round(v, 2)}" for k, v in physiology_data.items() if pd.notna(v)])

            # Retrieve activity features
            location_change_count = row['location_change_count']
            outside_time = row['outside_time']

            # Extract only valid location counts (excluding "location_change_count" itself)
            location_columns = [col for col in activity_features_df.columns if col.endswith('_count') and col != "location_change_count"]
            location_counts = "\n".join([f"- {col.replace('_count', ' Count')}: {row[col]}" for col in location_columns])

            # Construct structured text prompt
            text_prompt = f"""
            - **Age Group**: {age_group}
            - **Gender**: {gender}
            ---------------------
            - **Physiological Data**:
            {physiology_text if physiology_text else "- No physiological data available"}
            ---------------------
            - **Activity Information**:
            - Location Change Count: {location_change_count}
            - Time Spent Outside: {outside_time} min
            {location_counts}
            ---------------------
            Label: {mask_label}
            """
            # Prompt for stats
            if stats is not None:
              stats_prompt = "Average of features over normal days:\n"

              if patient_id in stats:
                for k,v in stats[patient_id].items():
                  stats_prompt += f"- {k}: {v}\n"
                stats_prompt+="---------------------\n"
              else:
                stats_prompt +="No prior stats available for this patient.\n"
              text_prompt = stats_prompt + text_prompt


            # Remove excessive spaces for formatting
            text_prompt = "\n".join([line.strip() for line in text_prompt.split("\n") if line.strip()])

            # Add the structured text to the respective list
            if split == "Train":
                sample_list.append(text_prompt.replace("{LABEL}", agitation_label))  # Replace {LABEL} in train
            else:
                sample_list.append(text_prompt)  # Test samples keep "[MASK]"
                label_list.append(agitation_label)  # Store true label for evaluation

    return train_samples, test_samples, test_labels


# Example usage:
train_samples, test_samples, test_labels = generate_balanced_text_prompts(demographics_df,
                            agit_df, physiology_df, activity_features_df,stats)

# Verify correct train-test split
print(f"Train Samples: {len(train_samples)} (Expected: 250 or max possible)")
print(f"Test Samples: {len(test_samples)} (Expected: 20 or max possible)")
print(f"Stored Test Labels: {len(test_labels)} (should match test samples)")

# Display sample outputs
print("\n===== TRAIN SAMPLE (with embedded labels) =====")
print(train_samples[0])  # Properly formatted for zero-shot inference

print("\n===== TEST SAMPLE (without labels) =====")
print(test_samples[0])  # Testing samples don't have a label

print("\n===== TEST LABELS (Stored for Evaluation) =====")
print(test_labels[:])  # Show a few stored true labels
```

```python
def get_few_shot_prompt(example_prompts_list, question_prompt, question=None, context=None, instruction=None, knowledge=None,
                        INSTRUCTION_LABEL='Instruction: ',
                        CONTEXT_LABEL='Context: ',
                        QUESTION_LABEL='Question: ',
                        ):
    """
    Constructs a few-shot learning prompt.

    Parameters:
    - example_prompts_list: List of example prompts used for in-context learning.
    - question_prompt: A prompt without an answer.
    - question: (Optional) A separate question to be included.
    - instruction: (Optional) Additional instruction.
    - knowledge: (Optional) Additional background knowledge.
    """
    few_shot_prompt = ""

    if instruction:
      few_shot_prompt = INSTRUCTION_LABEL + instruction + "\n\n"

    # Add context information
    if context:
      few_shot_prompt += CONTEXT_LABEL + context + "\n\n"

    # Construct few-shot examples
    if example_prompts_list:
      few_shot_prompt += "\n\n".join([f"Example {i}:\n{p}" for i, p in enumerate(example_prompts_list)])+ "\n\n"

    # Add question prompts
    if question:
      few_shot_prompt += QUESTION_LABEL + question + "\n"

    few_shot_prompt += question_prompt

    return few_shot_prompt
def get_prompt_label(prompt):
  return prompt.split("\n")[-1].split("Label:")[-1].strip()
```

```python
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, confusion_matrix, roc_auc_score, precision_recall_curve, auc

def evaluate(y_pred,y_true):
    # Calculate Accuracy
    accuracy = accuracy_score(y_true, y_pred)

    # Calculate F1 Score
    f1 = f1_score(y_true, y_pred)

    # Calculate Precision
    precision = precision_score(y_true, y_pred)

    # Calculate Sensitivity (Recall)
    sensitivity = recall_score(y_true, y_pred)

    # Calculate Specificity
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    specificity = tn / (tn + fp)

    # Calculate ROC AUC
    roc_auc = roc_auc_score(y_true, y_pred)

    # Calculate PR AUC
    precision_curve, recall_curve, _ = precision_recall_curve(y_true, y_pred)
    pr_auc = auc(recall_curve, precision_curve)

    return accuracy, f1, precision, sensitivity, specificity, roc_auc, pr_auc

y_pred = [1 if p=='Agitated' else 0 for p in copilot_50_2]
y_true = [1 if p=='Agitated' else 0 for p in test_labels]
accuracy, f1, precision, sensitivity, specificity, roc_auc, pr_auc = evaluate(y_pred,y_true)
print(f"Accuracy: {accuracy}")
print(f"F1 Score: {f1}")
print(f"Precision: {precision}")
print(f"Sensitivity (Recall): {sensitivity}")
print(f"Specificity: {specificity}")
print(f"ROC AUC: {roc_auc}")
print(f"PR AUC: {pr_auc}")
```

```python
# Define the function to process train samples efficiently
def process_all_train_samples(train_samples, model, tokenizer, max_tokens=100):
    """
    Feeds all training samples into GPT-2 without printing intermediate outputs.
    Only shows progress updates.
    """
    num_samples = len(train_samples)

    for i, sample in enumerate(train_samples):
        if i % 5 == 0:  # Print progress every 5 samples
            print(f"Processing train sample {i+1}/{num_samples}...")

        # Tokenize input
        input_ids = tokenizer.encode(sample, return_tensors="pt")

        # Generate output (GPT-2 processes the sample)
        with torch.no_grad():
            model.generate(input_ids, max_new_tokens=max_tokens, pad_token_id=tokenizer.eos_token_id)

    print("\n===== Training Completed! All samples processed. =====")

# Run the function on all train samples
process_all_train_samples(train_samples, model, tokenizer)
```

```python
import re
import torch

def generate_gpt2_prediction(prompt, max_length=250):
    """Generates GPT-2 output given a test sample prompt."""
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512)
    with torch.no_grad():
        outputs = model.generate(**inputs, max_length=max_length, num_return_sequences=1)

    generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

    return generated_text

def extract_predicted_label(generated_text):
    """Extracts the predicted label (Agitated/Normal) from GPT-2's generated response."""
    match = re.search(r'Label(?: \(Choose Agitated or Normal\))?:?\s*(Agitated|Normal)', generated_text, re.IGNORECASE)
    if match:
        return match.group(1).capitalize()  # Ensure consistent formatting
    return "Unknown"

# Modify test samples format to ensure GPT-2 understands the task better
test_samples = [sample.replace("Label: [MASK]",
    "Based on the above information, is the patient Agitated or Normal? Choose one.\nPrediction:")
    for sample in test_samples]

# Initialize counters
correct_predictions = 0
total_samples = len(test_samples)
predictions = []

print("\n===== Running GPT-2 on Test Samples =====")

for idx, (prompt, true_label) in enumerate(zip(test_samples, test_labels)):
    print(f"\n🔹 **Testing Sample {idx+1}/{total_samples}**")
    print("🚀 Feeding prompt to GPT-2...")

    # Generate GPT-2 prediction
    generated_text = generate_gpt2_prediction(prompt)

    # Extract predicted label
    predicted_label = extract_predicted_label(generated_text)
    predictions.append(predicted_label)

    # Compare with ground truth
    is_correct = predicted_label == true_label
    correct_predictions += int(is_correct)

    # Display result
    print(f"📝 **Predicted Label:** {predicted_label} | ✅ **True Label:** {true_label} {'✔' if is_correct else '❌'}")
    print("--------------------------------------------------")

# Compute accuracy
accuracy = (correct_predictions / total_samples) * 100
print(f"\n🎯 **Final Accuracy:** {accuracy:.2f}% ({correct_predictions}/{total_samples})")

# Display summary of predictions
print("\n===== Prediction Summary =====")
for i in range(total_samples):
    print(f"Test Sample {i+1}: Predicted={predictions[i]} | True={test_labels[i]}")
```
