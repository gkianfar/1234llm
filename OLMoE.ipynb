```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_id = "allenai/OLMoE-1B-7B-0125"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",             # Optional: to place on correct device
    load_in_8bit=True              # or use `load_in_4bit=True` for more compression
)

```

```python
execution_time_per_length = {}
```

```python
import time

max_length = 100

start = time.time()

inputs = tokenizer("List a number of challenges of running LLMs on Colab.", return_tensors="pt").to(model.device)
#print(tokenizer.decode(outputs[0], skip_special_tokens=True))

out = model.generate(
    **inputs,
    #max_new_tokens=max_length,
    top_p=1,              # Or use nucleus sampling
    temperature=0.7,         # Controls randomness
    repetition_penalty=1.2,  # Penalize repetition
)
print(tokenizer.decode(out[0]))
# > # Bitcoin is a digital currency that is created and held electronically. No one controls it. Bitcoins aren't printed, like dollars or euros – they're produced by people and businesses running computers all around the world, using software that solves mathematical
end = time.time()

# Print the execution time
execution_time_per_length[max_length] = end - start
print(f"Execution time: {end - start:.4f} seconds")
```
